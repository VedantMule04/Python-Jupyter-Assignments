```python
def student_info():
    student = ("Abhii", 20, "A")
    print("Name:", student[0])
    print("Age:", student[1])
    print("Grade:", student[2])

student_info()
```

```python
def grocery_prices():
    prices = (25, 50, 30, 45)
    total = sum(prices)
    print("Grocery Prices:", prices)
    print("Total Price:", total)

grocery_prices()
```

```python
def item_price_pair():
    items = (("Milk", 25), ("Eggs", 50), ("Bread", 20))
    for item in items:
        print(f"{item[0]} - ₹{item[1]}")

item_price_pair()
```

```python
def train_schedule():
    schedule = (("Rajdhani", "10:00"), ("Shatabdi", "12:30"), ("Duronto", "17:00"))
    for train in schedule:
        print(f"{train[0]} departs at {train[1]}")

train_schedule()
```

```python
def sort_employees():
    employees = (("John", 40000), ("Alice", 55000), ("Raj", 30000))
    sorted_employees = sorted(employees, key=lambda x: x[1], reverse=True)
    for emp in sorted_employees:
        print(f"{emp[0]} - ₹{emp[1]}")

sort_employees()
```

```python
def count_high_scores():
    marks = (67, 88, 92, 74, 76, 55)
    count = sum(1 for mark in marks if mark > 75)
    print("Students scoring above 75:", count)

count_high_scores()
```

```python
def avg_temperature():
    temperatures = (29.5, 30.0, 32.2, 31.5, 28.9)
    average = sum(temperatures) / len(temperatures)
    print("Average Temperature:", round(average, 2))

avg_temperature()
```

```python
def contact_info():
    contacts = (("Anil", "9876543210"), ("Priya", "9123456780"))
    for contact in contacts:
        print(f"Name: {contact[0]}, Phone: {contact[1]}")

contact_info()
```

```python
def exam_schedule():
    exams = (("Math", "9:00 AM"), ("Science", "11:30 AM"), ("English", "2:00 PM"))
    for exam in exams:
        print(f"{exam[0]} exam at {exam[1]}")

exam_schedule()
```
