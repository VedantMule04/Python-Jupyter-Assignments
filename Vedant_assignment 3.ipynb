```python
def create_set():
    my_set = {1, 2, 3, 4, 5}
    print("Set elements:", my_set)

create_set()
```

```python
def add_element():
    my_set = {1, 2, 3}
    my_set.add(4)
    print("After adding:", my_set)

add_element()
```

```python
def remove_element():
    my_set = {1, 2, 3, 4}
    my_set.remove(3)
    print("After removing 3:", my_set)

remove_element()
```

```python
def clear_set():
    my_set = {10, 20, 30}
    my_set.clear()
    print("Set after clearing:", my_set)

clear_set()
```

```python
def copy_set():
    set1 = {1, 2, 3}
    set2 = set1.copy()
    print("Copied Set:", set2)

copy_set()
```

```python
def check_element():
    my_set = {10, 20, 30}
    print(20 in my_set)

check_element()
```

```python
def set_length():
    my_set = {1, 2, 3, 4, 5}
    print("Length of set:", len(my_set))

set_length()
```

```python
def iterate_set():
    my_set = {"apple", "banana", "cherry"}
    for item in my_set:
        print(item)

iterate_set()
```

set operation

```python
def union_sets():
    a = {1, 2, 3}
    b = {3, 4, 5}
    print("Union:", a | b)

union_sets()
```

```python
def intersection_sets():
    a = {1, 2, 3}
    b = {2, 3, 4}
    print("Intersection:", a & b)

intersection_sets()
```

```python
def difference_sets():
    a = {1, 2, 3}
    b = {2, 3, 4}
    print("Difference (a - b):", a - b)

difference_sets()
```

```python
def symmetric_diff_sets():
    a = {1, 2, 3}
    b = {2, 3, 4}
    print("Symmetric Difference:", a ^ b)

symmetric_diff_sets()
```

```python
def is_subset():
    a = {1, 2}
    b = {1, 2, 3}
    print("Is subset:", a.issubset(b))

is_subset()
```

```python
def is_superset():
    a = {1, 2, 3}
    b = {1, 2}
    print("Is superset:", a.issuperset(b))

is_superset()
```

```python
def is_disjoint():
    a = {1, 2}
    b = {3, 4}
    print("Are disjoint:", a.isdisjoint(b))

is_disjoint()
```

```python
def remove_duplicates():
    lst = [1, 2, 2, 3, 4, 4, 5]
    unique = set(lst)
    print("Without duplicates:", unique)

remove_duplicates()
```

```python
def list_to_set_to_list():
    lst = [1, 2, 2, 3]
    s = set(lst)
    new_list = list(s)
    print("New list:", new_list)

list_to_set_to_list()
```

```python
def common_elements():
    a = [1, 2, 3, 4]
    b = [3, 4, 5, 6]
    print("Common elements:", set(a) & set(b))

common_elements()
```

```python
def difference_lists():
    a = [1, 2, 3, 4]
    b = [3, 4, 5, 6]
    print("In a not in b:", set(a) - set(b))

difference_lists()
```

```python
def set_comprehension():
    squares = {x**2 for x in range(1, 6)}
    print("Squares:", squares)

set_comprehension()
```
