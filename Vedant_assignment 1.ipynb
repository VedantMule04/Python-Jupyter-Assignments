```python
def count_digits(num):
    count = 0
    while num != 0:
        num = num // 10
        count += 1
    print("Total digits:", count)

# Example usage
count_digits(12345)
```

```python
def reverse_number_pattern(n):
    for i in range(n, 0, -1):
        for j in range(i, 0, -1):
            print(j, end=" ")
        print()

# Example usage
reverse_number_pattern(5)
```

```python
def print_primes(start, end):
    for num in range(start, end + 1):
        if num > 1:
            for i in range(2, int(num ** 0.5) + 1):
                if num % i == 0:
                    break
            else:
                print(num, end=" ")

# Example usage
print_primes(10, 50)
```

```python
def is_palindrome(num):
    original = str(num)
    reversed_num = original[::-1]
    if original == reversed_num:
        print(f"{num} is a Palindrome")
    else:
        print(f"{num} is not a Palindrome")

# Example usage
is_palindrome(121)
```

```python
def is_armstrong(num):
    power = len(str(num))
    total = sum(int(digit) ** power for digit in str(num))
    if total == num:
        print(f"{num} is an Armstrong number")
    else:
        print(f"{num} is not an Armstrong number")

# Example usage
is_armstrong(153)
```

```python
def find_maximum(a, b, c):
    maximum = max(a, b, c)
    print("Maximum number is:", maximum)

# Example usage
find_maximum(10, 20, 15)
```

```python
def sum_of_digits(num):
    total = 0
    while num > 0:
        total += num % 10
        num = num // 10
    print("Sum of digits is:", total)

# Example usage
sum_of_digits(1234)
```

```python
def print_summation_pattern(n):
    total = 0
    for i in range(1, n + 1):
        total += i
        expression = " + ".join(str(x) for x in range(1, i + 1))
        print(f"{expression} = {total}")

# Example usage
print_summation_pattern(5)
```
